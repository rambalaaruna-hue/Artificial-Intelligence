```python
import pandas as pd
data = {
    "Area":[600, 800, 1000, 1200,1500, 1800, 2000, 2200, 2500, 2800],
    "Price":[30,40,50,58,72,85,95,105,118,130]
}

df = pd.DataFrame(data)

print(df)
```

```python

# 3. Separate features and target
X = df["Area"]
y = df["Price"]
```

```python
x_mean = X.mean()
y_mean = y.mean()

print("Mean of X:" , x_mean)
print("Mean of Y:" , y_mean)
```

```python
# 1. Calculate the numerator and denominator for the slope formula
numerator = ((X - x_mean) * (y - y_mean)).sum()
denominator = ((X - x_mean) ** 2).sum()

# 2. Fix the typo: change 'numenator' to 'numerator'
m = numerator / denominator

# 3. Print the final calculated slope
print(f"Slope (m): {m:.5f}")
```

```python
c = y_mean - (m * x_mean)
print("Intercept (c):", c)
```

```python

predicted_price = m * X + c
df["Predicted_Price"] = predicted_price
print(df)
```

```python
new_area = 1600
predicted = m * new_area + c
print(f"Predicted price for {new_area} sq.ft house: {predicted:}")
print(df)
```

```python
# Fix: Capitalize 'Price' and add the underscore to 'Predicted_Price'
df["Error"] = df["Price"] - df["Predicted_Price"]
print(df)
```

```python
ss_res = ((y - predicted_price) ** 2).sum()
ss_tot = ((y - y.mean()) ** 2).sum()
r2_score = 1 - (ss_res / ss_tot)
print("R2 Score:", round(r2_score, 4))
```

```python
weights = {
    "slope": m,
    "intercept": c
}

print(weights)
```
