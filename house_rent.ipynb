```python
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.svm import SVR
import sklearn.metrics as metrics

# 1. RUN THE SPLIT using your variables (x_df and y from earlier steps)
# test_size=0.20 and random_state=40 as per your example
x_train, x_test, y_train, y_test = train_test_split(x_df, y, test_size=0.20, random_state=40)

# 2. Train your SVR model
model = SVR()
model.fit(x_train, y_train)

# 3. Predict safely (x_test is now properly defined in memory)
y_pred = model.predict(x_test)

# 4. Evaluate using the metrics module you imported
mae = metrics.mean_absolute_error(y_test, y_pred)
mse = metrics.mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = metrics.r2_score(y_test, y_pred)

# 5. Print results
print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R2 Score:", r2)

```
